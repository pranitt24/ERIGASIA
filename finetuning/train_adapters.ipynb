```python
import os
# Disables Triton to prevent the Windows crash
os.environ["UNSLOTH_COMPILE_DISABLE"] = "1"

# Unsloth MUST be imported before torch
from unsloth import FastLanguageModel
import torch

max_seq_length = 2048 # Tuned for your BKT assignments

# Load Base Model
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "Qwen/Qwen2.5-7B-Instruct",
    max_seq_length = max_seq_length,
    dtype = None,
    load_in_4bit = True, # Uses 4-bit quantization to easily fit in your 24GB VRAM
)

# Apply LoRA Adapters
model = FastLanguageModel.get_peft_model(
    model,
    r = 16, # Choose any number > 0 ! Suggested 8, 16, 32, 64, 128
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj",],
    lora_alpha = 16,
    lora_dropout = 0, 
    bias = "none",    
    use_gradient_checkpointing = "unsloth",
    random_state = 3407,
    use_rslora = False,
    loftq_config = None,
)
```
